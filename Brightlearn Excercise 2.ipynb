TASK 02

```python
df['unit_price'] = (df['unit_price'].str.replace(',', '.').astype(float))
```

```python
df['revenue']=df['unit_price']*df['transaction_qty']
df.head(5)
```

```python
df[['unit_price', 'transaction_qty', 'revenue']].head(5)
df['revenue'].dtype
```

```python
df[['unit_price', 'transaction_qty', 'revenue']].head()
```

```python
df[['unit_price', 'transaction_qty']].dtypes
```

```python
df["transaction_time"] = pd.to_datetime(df["transaction_time"])
df["hour"] = df["transaction_time"].dt.hour
df.sample(5)
```

```python
df["transaction_date"] = pd.to_datetime(df["transaction_date"])
df["Month"] = df["transaction_date"].dt.month
df.sample(5)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='unit_price', kde=True)

plt.title('Distribution of Unit Price')
plt.xlabel('Unit Price')
plt.ylabel('Frequency')
plt.show()
```

The shape is skew right

```python
revenue_by_store = (
    df.groupby('store_location', as_index=False)['revenue']
      .sum()
)

plt.figure(figsize=(8, 5))
sns.barplot(data=revenue_by_store, x='store_location', y='revenue', color='maroon', width=0.8)

plt.title('Total Revenue by Store Location')
plt.xlabel('Store Location')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
hourly_transactions = df.groupby('hour').size().reset_index(name='transactions')

plt.figure(figsize=(10, 5))
sns.barplot(data=hourly_transactions, x='hour', y='transactions')

plt.title('Transactions by Hour of Day')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Transactions')
plt.show()
```

The store is busiest around 10 am

TASK 03

```python
sns.heatmap(df[["unit_price","transaction_qty","revenue"]].corr(numeric_only=True, method = "spearman"), annot=True, fmt=".2f", cmap="coolwarm")
```

```python
pivot_table=pd.pivot_table(df,
               values="revenue",
               index="store_location",
               columns="product_category",
               aggfunc="sum",
               fill_value=0)

display(pivot_table)
```

```python
pivot_table.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Revenue by Store Location and Product Category')
plt.xlabel('Store Location')
plt.ylabel('Revenue')
plt.legend(title='Product Category')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
revenue_by_category=df.groupby('product_category')['revenue'].sum().sort_values(ascending=False)
print(revenue_by_category)
```

Coffee generates the most revenue

```python
revenue_by_store=(df.groupby('store_location')['revenue'].sum()).sort_values(ascending=False)
print(revenue_by_store)
```

Hells kitchen generates the most revenue

```python
df["product_category"].value_counts().plot.bar(color='red')
plt.show()
```

```python
cat_rev = df.groupby("product_category")["revenue"].sum().reset_index()

cat_rev.columns = ["category","revenue"]

cat_rev = cat_rev.sort_values("revenue",ascending = False)

cat_rev
```

```python
sns.barplot(data=cat_rev, x="category", y="revenue")
plt.xticks(rotation=30, ha="right")
plt.show()
```

```python
import plotly.express as px
```

```python
fg = px.bar(cat_rev, x="category", y="revenue",
            color="revenue",
            # color_continuous_scale="Greens"
            )
fg.show()
```
